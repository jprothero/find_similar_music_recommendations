```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# def npmi(px, py, pxy, eps=1e-400):
#     px = px + eps
#     py = py + eps
#     pxy = pxy + eps

def npmi(px, py, pxy, λ=0):
    assert px > 0
    assert py > 0
    assert pxy >= 0
    
    assert px <= 1
    assert py <= 1
    assert pxy <= 1
    
    assert pxy <= px
    assert pxy <= py
    
    assert λ >= 0
    
    if pxy == 0:
        return -1
    else:
        if λ > 0:
            pxy2 = min(pxy + λ, 1)
            px2 = min(px + np.sqrt(λ), 1)
            py2 = min(py + np.sqrt(λ), 1)
            
#             pxy2 = min(pxy + λ**2, 1)
#             px2 = min(px + λ**2, 1)
#             py2 = min(py + λ**2, 1)
            
            return (np.log2(pxy2) - (np.log2(px2) + np.log2(py2))) / -np.log2(pxy2)
        else:
            return (np.log2(pxy) - (np.log2(px) + np.log2(py))) / -np.log2(pxy)
```

```python
n = 256

p = 1/2

px = p
py = p

# simplifying assumption
assert py >= px

x = np.linspace(0, px, n)
    
for λ in np.linspace(0, 1):
    plt.plot(x, [npmi(px, py, pxy, λ) for pxy in x]);
```
